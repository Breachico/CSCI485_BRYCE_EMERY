## 1. K-means vs DBSCAN

### Classify Data with Non-Convex Shapes

A **convex shape** is one where:

> For any two points inside the shape, the straight line between them lies entirely within the shape.

Think: circles, ellipses, blobs — geometrically well-behaved.

A **non-convex shape**, by contrast, is more... rebellious.

Examples:
- A crescent moon  
- A horseshoe  
- Two interlocking rings  
- The classic “two moons” dataset

These shapes have **indentations or internal gaps**, so that some straight lines between points inside the cluster would **exit and re-enter** the cluster.

---

### Why Does K-means Struggle with Non-Convex Shapes?

K-means is built on **Euclidean distance** and **centroid averaging**, which comes with hidden assumptions:

1. **Spherical Assumption**  
   K-means draws boundaries like perfect little soap bubbles — it assumes that clusters are **globular** and **equidistant** from a center point.

2. **Linear Separation**  
   K-means can only separate clusters using **linear boundaries**. It can’t bend, twist, or wrap.

3. **Mean as Anchor**  
   The centroid (mean) pulls everything toward the center, regardless of how the shape of the actual cluster behaves.

So when clusters are **non-convex** — say, shaped like moons or spirals — K-means will:

- **Misclassify boundary points**, pulling them toward the wrong centroid
- **Split up one curved cluster** into multiple parts
- **Lump separate parts** of a U-shape into the same cluster if the centroids are misleading

---

### Philosophical Bottom Line:

K-means lives in a **Euclidean world** of symmetry and balance.  
Non-convex data lives in a **topological world** of twists and contours.

To use K-means well, the data must **fit its worldview**.

Otherwise, you're forcing square centroids into round moons.

---

### Visual Example

**Two Moons Dataset**

- DBSCAN: Captures the two moons cleanly  
- K-means: Splits each moon in half due to straight-line boundaries  


```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.cluster import KMeans, DBSCAN

# Step 1: Generate non-convex (two moons) data
X, y_true = make_moons(n_samples=300, noise=0.05, random_state=42)
# Try with bigger noise value (<0.2)

# Step 2: Apply K-means
kmeans = KMeans(n_clusters=2, random_state=42)
kmeans_labels = kmeans.fit_predict(X)

# Step 3: Apply DBSCAN
dbscan = DBSCAN(eps=0.2, min_samples=5)
dbscan_labels = dbscan.fit_predict(X)

# Step 4: Plot the results side-by-side
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# K-means plot
ax1.scatter(X[:, 0], X[:, 1], c=kmeans_labels, cmap='viridis', s=30)
ax1.set_title("K-means Clustering")
ax1.set_xlabel("Feature 1")
ax1.set_ylabel("Feature 2")

# DBSCAN plot
ax2.scatter(X[:, 0], X[:, 1], c=dbscan_labels, cmap='viridis', s=30)
ax2.set_title("DBSCAN Clustering")
ax2.set_xlabel("Feature 1")
ax2.set_ylabel("Feature 2")

plt.tight_layout()
plt.show()
```

## 2. K-means vs GMM

* Creates two extreme elliptical clusters, tilted in different directions.
  * K-means slices through them as if they're spherical — messy and wrong.
  * GMM:
    * Learns the correct orientation and shape
    * Overlays red ellipses showing its probabilistic model
    * Provides elegant, curved boundaries where needed

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.mixture import GaussianMixture
from sklearn.cluster import KMeans
from matplotlib.patches import Ellipse

def draw_ellipse(position, covariance, ax, **kwargs):
    if covariance.shape == (2, 2):
        U, s, _ = np.linalg.svd(covariance)
        angle = np.degrees(np.arctan2(U[1, 0], U[0, 0]))
        width, height = 2 * np.sqrt(s)
    else:
        angle = 0
        width, height = 2 * np.sqrt(covariance)

    for nsig in range(1, 4):
        ellipse = Ellipse(
            xy=position,
            width=nsig * width,
            height=nsig * height,
            angle=angle,
            fill=False,
            linewidth=2,
            **kwargs
        )
        ax.add_patch(ellipse)

# Step 1: Generate highly elliptical data
np.random.seed(42)
mean1 = [0, 0]
cov1 = [[10, 8], [8, 10]]  # Strong tilt, highly elongated
data1 = np.random.multivariate_normal(mean1, cov1, 200)

#mean2 = [15, 10]
#cov2 = [[9, -7], [-7, 9]]  # Opposite tilt
mean2 = [-2, 10]
cov2 = [[10, 8], [8, 10]]  # Strong tilt, highly elongated
data2 = np.random.multivariate_normal(mean2, cov2, 200)

X = np.vstack([data1, data2])

# Step 2: Apply K-means
kmeans = KMeans(n_clusters=2, random_state=0)
kmeans_labels = kmeans.fit_predict(X)

# Step 3: Apply GMM
gmm = GaussianMixture(n_components=2, covariance_type='full', random_state=0)
gmm.fit(X)
gmm_labels = gmm.predict(X)

# Step 4: Plot results
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# K-means plot
ax1.scatter(X[:, 0], X[:, 1], c=kmeans_labels, cmap='viridis', s=30)
ax1.set_title("K-means Clustering (Fails on Ellipses)")
ax1.set_xlabel("Feature 1")
ax1.set_ylabel("Feature 2")

# GMM plot
ax2.scatter(X[:, 0], X[:, 1], c=gmm_labels, cmap='viridis', s=30)
ax2.set_title("GMM Clustering (Adapts to Elliptical Shape)")
ax2.set_xlabel("Feature 1")
ax2.set_ylabel("Feature 2")

# Draw GMM ellipses
for pos, covar in zip(gmm.means_, gmm.covariances_):
    draw_ellipse(pos, covar, ax2, edgecolor='red')

plt.tight_layout()
plt.show()
```

## Visual Embodiment of the Gaussian Distribution’s Inner Geometry

### What Do the Contours Represent?

The ellipses drawn around each Gaussian component in a GMM are **level sets** (or **isocontours**) of the **multivariate Gaussian probability density function (PDF)**.

In plain terms:

> They represent regions where the probability density is **constant** — like topographic rings on a landscape of likelihood.

---

### Mathematically Speaking...

A **2D Gaussian** has the form:

> \$
f(x) = \frac{1}{2\pi |\Sigma|^{1/2}} \exp\left( -\frac{1}{2} (x - \mu)^T \Sigma^{-1} (x - \mu) \right)
\$

Where:
- $ \mu $: mean (center of the cluster)
- $ \Sigma $: covariance matrix (shape, orientation, spread)
- $ x $: any point in the space

Now here's the magic:

> \$
(x - \mu)^T \Sigma^{-1} (x - \mu) = c
\$

This equation defines an **ellipse** — it's the **Mahalanobis distance** squared between $ x $ and $ \mu $, adjusted by the shape of $ \Sigma $.

---

### What Each Ellipse Means

Each ellipse corresponds to a specific value of $ c $ — a **probability density contour**:

- $ c = 1 $: **1 standard deviation** → ~68% of the data falls inside
- $ c = 4 $: **2 std deviations** → ~95%
- $ c = 9 $: **3 std deviations** → ~99.7%

That’s why in the code we loop over `nsig = 1, 2, 3` — each is one more standard deviation of the Gaussian.

---

### Geometric Insight

- The **orientation** of the ellipse is determined by the **eigenvectors** of $ \Sigma $
- The **length of the axes** is proportional to the **square roots of the eigenvalues**
- The ellipse says: “This is the region where points are *likely* to occur under this Gaussian.”

---

### Summary Table

| Visual Element | Mathematical Meaning                          | Intuition |
|----------------|-----------------------------------------------|-----------|
| Ellipse center | $ \mu $ (mean)                              | Cluster center |
| Axis directions | Eigenvectors of $ \Sigma $                 | Tilt / orientation |
| Axis lengths   | $ \sqrt{\lambda_i} $ from eigenvalues       | Spread in each direction |
| Ellipse shape  | Level set of Mahalanobis distance             | Constant probability density |

---

### Philosophical Takeaway

> Those ellipses aren’t just decorative — they are **probabilistic borders**.  
> They whisper: *"Inside here, you’re likely; outside, you're a statistical outlier."*
